```python
fact_trips['month'] = fact_trips['date'].dt.to_period('M').dt.to_timestamp() #adding a month column
```

```python
monthly_city_trips = (fact_trips.groupby(['month', 'city_id']).agg(total_trips = ('trip_id' , 'count'), total_distance_travelled_km = ('distance_travelled(km)','sum'), total_fare = ('fare_amount', 'sum'), avg_passenger_rating = ('passenger_rating', 'mean'), avg_driver_rating = ('driver_rating', 'mean')).reset_index()) #creating a new table by aggregating columns and grouping by city_id and month
monthly_city_trips.head()
```

```python
master_table = monthly_city_trips.merge(city_target_passenger_rating, on = 'city_id', how = 'left') #joining targer_avg_passenger_rating column with monthly_city_trips
master_table.head()
```

```python
#merging fact_passenger_summary type table with master_table
master_table = pd.merge(master_table, fact_passenger_summary, on=['month', 'city_id'])
master_table.head()
```

```python
#merging monthly_target_new_passengers table with master_table
master_table = pd.merge(master_table, monthly_target_new_passengers, on= ['month', 'city_id'])
master_table.head()
```

```python
#Creating a new column to count target vs actual new passengers
master_table['new_passengers_variance'] = master_table['new_passengers'] - master_table['target_new_passengers']
```

```python
#merging monthly_target_trips table with master_table to have avg_passenger_rating
master_table = pd.merge(master_table, monthly_target_trips, on= ['month','city_id'])
```

```python
#Creating a new column trips_variance to count target vs actual trips
master_table['trips_variance'] = master_table['total_trips'] - master_table['total_target_trips']
```

```python
master_table.head()
```

```python
print(monthly_city_trips.shape)
print(monthly_city_trips.isna().sum())
```

```python
#Creating new column to count new and repeat passenger percentages
master_table['new_passenger_pct'] = master_table['new_passengers'] / master_table['total_passengers'] * 100
master_table['repeat_passenger_pct'] = master_table['repeat_passengers'] / master_table['total_passengers'] * 100
```

```python
#Rounding these newly created columns to two decimal numbers
master_table['new_passenger_pct'] = master_table['new_passenger_pct'].round(2)
master_table['repeat_passenger_pct'] = master_table['repeat_passenger_pct'].round(2)
```

```python
#Creating new columns to calculate percentages of new passenger variance and trip variance
master_table['new_pass_var_pct'] = master_table['new_passengers_variance'] / master_table['target_new_passengers'] * 100
```

```python
master_table['trips_var_pct'] = master_table['trips_variance'] / master_table['total_target_trips'] * 100
```

```python
#Rounding these newly created columns to two decimal numbers
master_table['new_pass_var_pct'] = master_table['new_pass_var_pct'].round(2)
master_table['trips_var_pct'] = master_table['trips_var_pct'].round(2)
```
